```python
subscriptions["is_churned"] = subscriptions["end_date"].notna()
```

```python
subscriptions["subscription_months"] = (
    subscriptions["subscription_days"] / 30
).round(1)
```

```python
payments["payment_month"] = payments["payment_date"].dt.to_period("M")
```

```python
payments["is_successful"] = payments["payment_status"].eq("Paid")
```

```python
# ==========================================
# FLOWLY BUSINESS EDA
# ==========================================

# 1. Customer overview
print("===== CUSTOMER OVERVIEW =====")
print("Total Customers:", customers["customer_id"].nunique())
print("Total Users:", users["user_id"].nunique())
print("Active Customers:", (~subscriptions["is_churned"]).sum())
print("Churned Customers:", subscriptions["is_churned"].sum())

# 2. Plan distribution
print("\n===== PLAN DISTRIBUTION =====")
print(subscriptions["plan"].value_counts())

# 3. Churn by plan
print("\n===== CHURN BY PLAN =====")

churn_by_plan = (
    subscriptions
    .groupby("plan")
    .agg(
        customers=("customer_id", "nunique"),
        churned=("is_churned", "sum"),
        churn_rate=("is_churned", "mean")
    )
)

churn_by_plan["churn_rate"] *= 100

print(churn_by_plan.sort_values("churn_rate", ascending=False))


# 4. Revenue by plan
print("\n===== REVENUE BY PLAN =====")

revenue_by_plan = (
    payments[payments["is_successful"]]
    .merge(
        subscriptions[["customer_id", "plan"]],
        on="customer_id",
        how="left"
    )
    .groupby("plan")["amount"]
    .sum()
    .sort_values(ascending=False)
)

print(revenue_by_plan)


# 5. Feature adoption
print("\n===== FEATURE ADOPTION =====")

feature_summary = (
    feature_usage
    .groupby("feature_name")
    .agg(
        total_usage=("usage_count", "sum"),
        usage_records=("usage_id", "count"),
        unique_customers=("customer_id", "nunique")
    )
    .sort_values("unique_customers", ascending=False)
)

feature_summary["adoption_rate"] = (
    feature_summary["unique_customers"]
    / customers["customer_id"].nunique()
    * 100
)

print(feature_summary)


# 6. Activity by event type
print("\n===== ACTIVITY EVENTS =====")
print(user_activity["event_type"].value_counts())


# 7. Support ticket categories
print("\n===== SUPPORT TICKETS =====")
print(support_tickets["category"].value_counts())


# 8. Average satisfaction
print("\n===== SUPPORT SATISFACTION =====")
print(
    "Average Satisfaction:",
    support_tickets["satisfaction_score"].mean()
)


# 9. Customer activity statistics
customer_activity = (
    user_activity
    .groupby("customer_id")
    .agg(
        total_events=("activity_id", "count"),
        total_session_minutes=("session_minutes", "sum"),
        avg_session_minutes=("session_minutes", "mean")
    )
)

print("\n===== CUSTOMER ACTIVITY =====")
print(customer_activity.describe())
```

```python
# ==========================================
# CUSTOMER INTELLIGENCE TABLE
# STEP 1 — CUSTOMER + SUBSCRIPTION
# ==========================================

customer_intel = customers.merge(
    subscriptions[
        [
            "customer_id",
            "plan",
            "start_date",
            "end_date",
            "seats",
            "monthly_price",
            "subscription_months",
            "is_churned"
        ]
    ],
    on="customer_id",
    how="left"
)

print("Shape:", customer_intel.shape)
print(customer_intel.head())
```

```python
# ==========================================
# STEP 2 — ACTIVITY METRICS
# ==========================================

activity_summary = (
    user_activity
    .groupby("customer_id")
    .agg(
        total_events=("activity_id", "count"),
        total_session_minutes=("session_minutes", "sum"),
        avg_session_minutes=("session_minutes", "mean"),
        active_days=("activity_date", "nunique")
    )
    .reset_index()
)

customer_intel = customer_intel.merge(
    activity_summary,
    on="customer_id",
    how="left"
)

print("Shape:", customer_intel.shape)
print(customer_intel.head())
```

```python
# ==========================================
# STEP 3 — FEATURE USAGE
# ==========================================

feature_summary_customer = (
    feature_usage
    .groupby("customer_id")
    .agg(
        total_feature_usage=("usage_count", "sum"),
        features_used=("feature_name", "nunique")
    )
    .reset_index()
)

customer_intel = customer_intel.merge(
    feature_summary_customer,
    on="customer_id",
    how="left"
)

print("Shape:", customer_intel.shape)
```

```python
# ==========================================
# STEP 4 — REVENUE
# ==========================================

revenue_summary = (
    payments[payments["is_successful"]]
    .groupby("customer_id")
    .agg(
        total_revenue=("amount", "sum"),
        successful_payments=("payment_id", "count")
    )
    .reset_index()
)

customer_intel = customer_intel.merge(
    revenue_summary,
    on="customer_id",
    how="left"
)

print("Shape:", customer_intel.shape)
```

```python
# ==========================================
# STEP 5 — SUPPORT
# ==========================================

support_summary = (
    support_tickets
    .groupby("customer_id")
    .agg(
        support_tickets=("ticket_id", "count"),
        avg_resolution_hours=("resolution_hours", "mean"),
        avg_satisfaction=("satisfaction_score", "mean")
    )
    .reset_index()
)

customer_intel = customer_intel.merge(
    support_summary,
    on="customer_id",
    how="left"
)

print("Shape:", customer_intel.shape)
```

```python
# ==========================================
# STEP 6 — HANDLE MISSING ANALYTICAL VALUES
# ==========================================

zero_columns = [
    "total_events",
    "total_session_minutes",
    "avg_session_minutes",
    "active_days",
    "total_feature_usage",
    "features_used",
    "total_revenue",
    "successful_payments",
    "support_tickets"
]

customer_intel[zero_columns] = (
    customer_intel[zero_columns].fillna(0)
)

print(customer_intel.isnull().sum())
```

```python
customer_intel.head()
```

```python
print("Final shape:", customer_intel.shape)
print("Customers:", customer_intel["customer_id"].nunique())
print("Columns:", list(customer_intel.columns))
```

```python
# ==========================================
# CHURN VS ENGAGEMENT ANALYSIS
# ==========================================

churn_analysis = (
    customer_intel
    .groupby("is_churned")
    .agg(
        customers=("customer_id", "nunique"),
        avg_events=("total_events", "mean"),
        avg_session_minutes=("total_session_minutes", "mean"),
        avg_active_days=("active_days", "mean"),
        avg_features_used=("features_used", "mean"),
        avg_feature_usage=("total_feature_usage", "mean"),
        avg_revenue=("total_revenue", "mean"),
        avg_support_tickets=("support_tickets", "mean"),
        avg_satisfaction=("avg_satisfaction", "mean")
    )
)

print(churn_analysis)
```

```python
churn_analysis.index = ["Active", "Churned"]

print(churn_analysis.round(2))
```

```python
# ==========================================
# NUMPY STATISTICAL ANALYSIS
# ==========================================

engagement = customer_intel["total_events"]

print("Mean:", np.mean(engagement))
print("Median:", np.median(engagement))
print("Standard Deviation:", np.std(engagement))

print("25th Percentile:", np.percentile(engagement, 25))
print("50th Percentile:", np.percentile(engagement, 50))
print("75th Percentile:", np.percentile(engagement, 75))
print("95th Percentile:", np.percentile(engagement, 95))
```

```python
# ==========================================
# STEP 4 — CUSTOMER ENGAGEMENT SEGMENTS
# ==========================================

q25 = np.percentile(customer_intel["total_events"], 25)
q75 = np.percentile(customer_intel["total_events"], 75)

print("25th Percentile:", q25)
print("75th Percentile:", q75)

customer_intel["engagement_level"] = np.select(
    [
        customer_intel["total_events"] <= q25,
        customer_intel["total_events"] > q75
    ],
    [
        "Low",
        "High"
    ],
    default="Medium"
)

print("\nEngagement Distribution:")
print(customer_intel["engagement_level"].value_counts())
```

```python
# ==========================================
# STEP 5 — ENGAGEMENT VS CHURN
# ==========================================

engagement_churn = (
    customer_intel
    .groupby("engagement_level")
    .agg(
        customers=("customer_id", "nunique"),
        churned_customers=("is_churned", "sum"),
        churn_rate=("is_churned", "mean")
    )
)

engagement_churn["churn_rate"] *= 100

print(engagement_churn.round(2))
```

```python
engagement_summary = (
    customer_intel
    .groupby("engagement_level")
    .agg(
        customers=("customer_id", "nunique"),
        avg_events=("total_events", "mean"),
        avg_session_minutes=("total_session_minutes", "mean"),
        avg_features_used=("features_used", "mean"),
        churn_rate=("is_churned", "mean")
    )
)

engagement_summary["churn_rate"] *= 100

print(engagement_summary.round(2))
```

```python
# ==========================================
# STEP 7 — REVENUE & CUSTOMER VALUE ANALYSIS
# ==========================================

revenue_analysis = (
    customer_intel
    .groupby("plan")
    .agg(
        customers=("customer_id", "nunique"),
        total_revenue=("total_revenue", "sum"),
        avg_revenue=("total_revenue", "mean"),
        median_revenue=("total_revenue", "median")
    )
    .sort_values("total_revenue", ascending=False)
)

print(revenue_analysis.round(2))
```

```python
revenue_churn = (
    customer_intel
    .groupby("is_churned")
    .agg(
        customers=("customer_id", "nunique"),
        avg_revenue=("total_revenue", "mean"),
        median_revenue=("total_revenue", "median")
    )
)

revenue_churn.index = ["Active", "Churned"]

print(revenue_churn.round(2))
```

```python
q25_revenue = np.percentile(customer_intel["total_revenue"], 25)
q75_revenue = np.percentile(customer_intel["total_revenue"], 75)

customer_intel["revenue_segment"] = np.select(
    [
        customer_intel["total_revenue"] <= q25_revenue,
        customer_intel["total_revenue"] > q75_revenue
    ],
    [
        "Low Value",
        "High Value"
    ],
    default="Medium Value"
)

print(customer_intel["revenue_segment"].value_counts())
```

```python
revenue_segment_churn = (
    customer_intel
    .groupby("revenue_segment")
    .agg(
        customers=("customer_id", "nunique"),
        churned_customers=("is_churned", "sum"),
        churn_rate=("is_churned", "mean"),
        avg_revenue=("total_revenue", "mean")
    )
)

revenue_segment_churn["churn_rate"] *= 100

print(revenue_segment_churn.round(2))
```

```python
# ==========================================
# STEP 11 — FEATURE ADOPTION ANALYSIS
# ==========================================

feature_adoption = (
    feature_usage
    .groupby("feature_name")
    .agg(
        total_usage=("usage_count", "sum"),
        unique_customers=("customer_id", "nunique"),
        usage_records=("usage_id", "count")
    )
    .reset_index()
)

feature_adoption["adoption_rate"] = (
    feature_adoption["unique_customers"]
    / customer_intel["customer_id"].nunique()
    * 100
)

feature_adoption = feature_adoption.sort_values(
    "adoption_rate",
    ascending=False
)

print(feature_adoption.round(2))
```

```python
# ==========================================
# STEP 12 — FEATURE USAGE VS CHURN
# ==========================================

customer_feature = (
    feature_usage
    .groupby("customer_id")
    .agg(
        total_feature_usage=("usage_count", "sum"),
        features_used=("feature_name", "nunique")
    )
    .reset_index()
)

feature_churn = customer_intel[
    ["customer_id", "is_churned"]
].merge(
    customer_feature,
    on="customer_id",
    how="left"
)

feature_churn[
    ["total_feature_usage", "features_used"]
] = feature_churn[
    ["total_feature_usage", "features_used"]
].fillna(0)

feature_churn_summary = (
    feature_churn
    .groupby("is_churned")
    .agg(
        customers=("customer_id", "nunique"),
        avg_feature_usage=("total_feature_usage", "mean"),
        avg_features_used=("features_used", "mean")
    )
)

feature_churn_summary.index = ["Active", "Churned"]

print(feature_churn_summary.round(2))
```

```python
# ==========================================
# STEP 13 — TOP FEATURES
# ==========================================

top_features = feature_adoption[
    [
        "feature_name",
        "adoption_rate",
        "total_usage"
    ]
].head(10)

print(top_features.round(2))
```

```python
# ==========================================
# STEP 14 — SUPPORT EXPERIENCE ANALYSIS
# ==========================================

support_summary = (
    support_tickets
    .groupby("customer_id")
    .agg(
        total_tickets=("ticket_id", "count"),
        avg_resolution_hours=("resolution_hours", "mean"),
        avg_satisfaction=("satisfaction_score", "mean")
    )
    .reset_index()
)

customer_support = customer_intel[
    ["customer_id", "is_churned"]
].merge(
    support_summary,
    on="customer_id",
    how="left"
)

print(customer_support.head())
```

```python
# ==========================================
# STEP 15 — SUPPORT VS CHURN
# ==========================================

support_churn = (
    customer_support
    .groupby("is_churned")
    .agg(
        customers=("customer_id", "nunique"),
        avg_tickets=("total_tickets", "mean"),
        avg_resolution_hours=("avg_resolution_hours", "mean"),
        avg_satisfaction=("avg_satisfaction", "mean")
    )
)

support_churn.index = ["Active", "Churned"]

print(support_churn.round(2))
```

```python
# ==========================================
# STEP 16 — SUPPORT CATEGORIES
# ==========================================

support_categories = (
    support_tickets
    .groupby("category")
    .agg(
        total_tickets=("ticket_id", "count"),
        avg_resolution_hours=("resolution_hours", "mean"),
        avg_satisfaction=("satisfaction_score", "mean")
    )
    .sort_values("total_tickets", ascending=False)
)

print(support_categories.round(2))
```
